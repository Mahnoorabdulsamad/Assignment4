## **01_add_two_numbers**


```python
def add_numbers():

  user:str = input("Enter your first number ")
  user:int = int(user)
  user2 = input("Enter your second number ")
  Add_numbers:int = int(user + int(user2))
  print(f"Add two numbers {Add_numbers} ")
if __name__ == "__main__":
  add_numbers()
```

# **03_fahrenheit_to_celsius**

```python
def main():
  fahrenheit_to_celsius = float(input("Enter temperature in Fahrenheit: "))
  degrees_celsius = (fahrenheit_to_celsius - 32) *  5.0/9.0
  print(f"Temperature in celsius is {degrees_celsius}")
if __name__ == "__main__":
  main()
```

# **05_triangle_perimeter**

```python
def main():
  side1:float = float(input("Enter the length of side 1: "))
  side2:float = float(input("Enter the length of side 2: "))
  side3:float = float(input("Enter the length of side 3: "))
  perimeter:float = float(side1 + side2 + side3)
  print(f"The perimeter of the triangle is  {perimeter}" )
if __name__ == "__main__":
  main()
```
